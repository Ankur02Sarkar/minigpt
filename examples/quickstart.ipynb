# minigpt_llm Quickstart

Generate text from a trained minigpt_llm checkpoint.


```python
import sys
import torch

sys.path.insert(0, '/opt/minigpt_llm')

# Load tokenizer
from tokenizers import Tokenizer
tokenizer = Tokenizer('/opt/minigpt_llm/tokenizer/vocab.json')

# Load model
from minigpt_llm.model.model import GPT, ModelConfig
config = ModelConfig.load('configs/minigpt-high.yaml')
model = GPT(config)
model.load_state_dict(torch.load('/opt/minigpt_llm/checkpoints/minigpt-high/best.pt', map_location='cpu'))
model.eval()

# Generate!
prompt = "Once upon a time"
input_ids = tokenizer.encode(prompt, add_special_tokens=False)['input_ids']
input_tensor = torch.tensor([input_ids])

with torch.no_grad():
    generated = model.generate(
        input_tensor,
        max_new_tokens=50,
        temperature=0.8,
        top_k=40,
        top_p=0.9
    )

output = tokenizer.decode(generated[0].tolist())
print(f'Prompt: {prompt}')
print(f'Generated: {output}')
```
